```python

import matplotlib.pyplot as plt

# Verified transportation costs
methods = ["VAM Initial Solution", "Optimized Solution"]
costs = [145380, 145140]

plt.figure(figsize=(9, 5))
bars = plt.bar(methods, costs, color=["#86A9D5", "#73B9A0"])

plt.title("Transportation Cost Comparison")
plt.ylabel("Total Transportation Cost (NGN)")
plt.ylim(0, 170000)

for bar, cost in zip(bars, costs):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        cost + 2000,
        f"NGN {cost:,.0f}",
        ha="center"
    )

plt.tight_layout()
plt.savefig("transportation_cost_comparison.png", dpi=300)
plt.show()
```
